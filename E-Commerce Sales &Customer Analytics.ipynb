```python
monthly_sales =(df.set_index('order_date').resample("ME")['Sales'].sum())
plt.figure(figsize=(10,5))
plt.plot(monthly_sales.index,monthly_sales.values,marker="o")
plt.xlabel('month')
plt.ylabel('Sales')
plt.title('Monthly Sales vs Trend')
plt.grid(True)
plt.tight_layout
plt.show()
```

```python
category_sales = df.groupby('category')['Sales'].sum()
plt.figure(figsize=(10,5))
plt.bar(category_sales.index,category_sales.values)
plt.xlabel('category')
plt.ylabel('Sales')
plt.title('TotalSales vs category')
plt.tight_layout
plt.show()
```

```python
profit_category = df.groupby('category')['profit'].sum().sort_values(ascending=False)
plt.figure(figsize=(10,5))
plt.bar(profit_category.index,profit_category.values)
plt.xlabel('category')
plt.ylabel('profit')
plt.title('Totalprofit by category')
plt.tight_layout
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['Sales'],df['profit'])
plt.xlabel('Sales')
plt.ylabel('profit')
plt.title('TotalSales vs profit')
plt.tight_layout
plt.show()
```

```python
satisfaction = df['customer_satisfaction'].value_counts() 
plt.figure(figsize=(10,5))
plt.pie(satisfaction.values,labels=satisfaction.index,autopct="%1.1f%%",startangle=90)
plt.title("customer_satisfaction") 
plt.show()
```

```python
region_sales = df.groupby("region")["Sales"].sum().sort_values(ascending=False)
plt.figure(figsize=(10,5))
region_sales.plot(kind="bar")
plt.title("Totalsales By Region")
plt.xlabel("region")
plt.ylabel("Sales")
plt.tight_layout
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(df['cost'])
plt.title("Total cost")
plt.tight_layout
plt.show()
```

```python
top_10_products = (df.groupby("product_name")["Sales"].sum().sort_values(ascending=False).head(10).sort_values())
plt.figure(figsize=(10,5))
top_10_products.plot(kind="barh")
plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")
plt.ylabel("product_name")
plt.tight_layout
plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x="Sales",y="profit",hue="category")
plt.title("Sales vs Profit")
plt.tight_layout    
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.lineplot(data=df, x="discount_pct",y="profit")
plt.title("discount_pct vs profit")
plt.tight_layout   
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x="sub_category",y="profit")
plt.title("sub_category vs profit")
plt.tight_layout   
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(data=df, x="quantity",bins=5)
plt.title("Total Quantity")
plt.tight_layout  
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.kdeplot(data=df, x="profit",hue="country",fill=True)
plt.title("Country Wise Profit")
plt.tight_layout  
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.violinplot(data=df, x="brand",y="profit",hue="brand")
plt.title("Brand Wise Profit")
plt.tight_layout  
plt.show()
```
